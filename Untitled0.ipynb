```python
import numpy as np
import matplotlib.pyplot as plt
# Initial conditions
g = 9.81
e = 0.8
ppf = 20
threshold = 0.1
position = [5.0, 10.0]
velocity = [0.0, 0.0]
gravity = [0.0, -g]
time = 0.0
times = [time]
positions = [position.copy()]
# Simulate each flight and bounce
while position[1] > 0 or velocity[1] > threshold:
    # Time until the ball reaches the ground
    t_hit = (
        velocity[1]
        + np.sqrt(velocity[1]**2 + 2*g*position[1])
    ) / g
    # Choose sample times during this flight
    local_time = np.linspace(0, t_hit, ppf + 1)
    # Calculate positions using a simple loop
    segment_positions = []
    for t in local_time:
        x = position[0] + velocity[0]*t + 0.5*gravity[0]*t**2
        y = position[1] + velocity[1]*t + 0.5*gravity[1]*t**2
        segment_positions.append([x, y])
    # Save times and positions without duplicating the start
    times.extend(time + local_time[1:])
    positions.extend(segment_positions[1:])
    # Update position at impact
    position = segment_positions[-1].copy()
    position[1] = 0.0
    # Update velocity at impact
    velocity[0] = velocity[0] + gravity[0]*t_hit
    velocity[1] = velocity[1] + gravity[1]*t_hit
    # Advance simulation time
    time += t_hit
    # Inelastic bounce
    velocity[1] = -e * velocity[1]
# Convert lists to NumPy arrays for plotting
times = np.array(times)
positions = np.array(positions)
# Plot the trajectory
plt.figure(figsize=(10, 5))
plt.plot(times, positions[:, 1])
plt.xlabel("Time (s)")
plt.ylabel("Height (m)")
plt.title("Inelastic Ball Bouncing")
plt.grid(True)
plt.show()
```

```python

import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML, display

# Create the scene
fig, ax = plt.subplots(figsize=(6, 7))

ax.set_xlim(0, 10)
ax.set_ylim(0, 11)
ax.set_xlabel("Horizontal position (m)")
ax.set_ylabel("Height (m)")
ax.set_title("Inelastic Collisions: Bouncing Ball")
ax.axhline(0, color="black", linewidth=2)
ax.grid(True)

# Create the ball
ball, = ax.plot([], [], "o", markersize=14)

# Text showing the current height
info = ax.text(0.05, 0.95, "", transform=ax.transAxes,
               verticalalignment="top")

def update(frame):
    x = positions[frame, 0]
    y = positions[frame, 1]

    ball.set_data([x], [y])
    info.set_text(f"Height: {y:.2f} m")

    return ball, info

ani = FuncAnimation(
    fig,
    update,
    frames=len(positions),
    interval=25,
    blit=True
)

plt.close(fig)
display(HTML(ani.to_jshtml()))
```
